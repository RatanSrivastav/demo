```python
x=['monday','tuesday','wednesday','thursday','friday']
y=[10,15,7,2,20]
```

```python
plt.plot(x,y,marker="o",color='green',linestyle=':',linewidth="2")
plt.title("bakery sales this week")
plt.xlabel("days")
plt.ylabel("no of items sold")
plt.grid()
plt.show()
```

#PYPLOT COMPLETE FUNCTIONS

```python
months=[1,2,3,4]
sales=[100,1500,1800,1200]
plt.plot(months,sales,color="blue",linestyle="--",linewidth="2",marker="o",label="SALES DATA",)
plt.legend(loc="lower right",fontsize=12) #or upper left
plt.xlabel("months")
plt.ylabel("no of item sold")
plt.grid(color="red",linestyle=":",linewidth="1")
plt.title("monthly sales")
#limit
plt.ylim(0,2000)
plt.xlim(0,3)
plt.xticks([ 1,2,3,4],['m1','m2','m3','m4'])

plt.show()
```

```python
#bar chart
```

```python
product=['A','B','C','D']
sales=[1000,1500,800,1200]
plt.bar(product,sales,color="orange",linewidth=2,label="sales data")
plt.xlabel("product")
plt.ylabel("sales")
plt.title("product sales comparison")
plt.legend()

plt.show()

# for horizontal write plt.barh
```

```python
#pie chart
```

```python
regions=['north','south','east','west']
sales=[1000,1500,800,1200]
plt.pie(sales, labels=regions, autopct='%1.1f%%', colors=['red','lightgreen','skyblue','coral'], wedgeprops=dict(width=1, edgecolor='black'))

plt.title("Revenue contribution By Regions")
plt.show()
```

```python
#histogram
```

```python
scores=[50,60,60,61,70,70,70,80,80,90,100,66,67,86,50]
plt.hist(scores,bins=5,color="green",linewidth=2,edgecolor="red")
plt.xlabel("score range")
plt.ylabel("no of student")
plt.grid()
plt.title("students score distribution")
plt.show()
```
